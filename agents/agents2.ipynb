```python
from google.genai import types

generation_chat_history = [
    types.Content(
        role="user",
        parts=[types.Part.from_text(text="Generate a Python implementation of the Merge Sort algorithm")]
    )
]
```

```python
config = types.GenerateContentConfig(
    system_instruction="You are a Python programmer tasked with generating high quality Python code."
    "Your task is to Generate the best content possible for the user's request. If the user provides critique," 
    "respond with a revised version of your previous attempt."
)

response = client.models.generate_content(
  contents = generation_chat_history,
  model = 'gemini-2.5-flash',
  config = config
)
mergesort_code = response.text
generation_chat_history.append(
  types.Content(
    role = 'assistant',
    parts = [types.Part.from_text(text = mergesort_code)]
  )
)
```

```python
reflection_chat_history = [
  types.Content(
    role = 'user',
    parts = [types.Part.from_text(text = mergesort_code)]
  )
]
config = types.GenerateContentConfig(
    system_instruction="You are Andrej Karpathy, an experienced computer scientist. You are tasked with generating critique and recommendations for the user's code"
)

critique = client.models.generate_content(
    contents = reflection_chat_history,
    model = 'gemini-2.5-flash',
    config = config
)
reflection_chat_history.append(
  types.Content(
    role = 'assistant',
    parts = [types.Part.from_text(text = critique.text)]
  )
)
display_markdown(critique.text, raw = True)
```

```python
import time
from google import genai
from google.genai import types
from google.genai.errors import ServerError, APIError
from tenacity import retry, stop_after_attempt, wait_exponential, retry_if_exception_type

client = genai.Client()

# This decorator automatically retries if a 503 ServerError occurs.
# It waits exponentially: 2s, 4s, 8s, up to 5 total attempts before giving up.
@retry(
    stop=stop_after_attempt(5),
    wait=wait_exponential(multiplier=1, min=2, max=10),
    retry=retry_if_exception_type((ServerError, APIError)),
    reraise=True
)
def safe_generate_content(model, contents, config):
    return client.models.generate_content(
        model=model,
        contents=contents,
        config=config
    )

def run_gemini_reflection_agent(user_msg, generation_system_prompt, reflection_system_prompt, n_steps=2):
    gen_config = types.GenerateContentConfig(system_instruction=generation_system_prompt)
    reflect_config = types.GenerateContentConfig(system_instruction=reflection_system_prompt)
    
    current_code = ""
    gen_history = [types.Content(role="user", parts=[types.Part.from_text(text=user_msg)])]
    reflect_history = []

    for step in range(n_steps):
        print(f"--- Step {step + 1}: Generating/Improving Code ---")
        
        # Use our new resilient call wrapper
        gen_response = safe_generate_content(
            model='gemini-2.5-pro',
            contents=gen_history,
            config=gen_config
        )
        current_code = gen_response.text
        gen_history.append(types.Content(role="model", parts=[types.Part.from_text(text=current_code)]))
        
        print(f"--- Step {step + 1}: Critiquing (Karpathy Reflection) ---")
        
        critique_prompt = f"Critique the following Python implementation of Merge Sort:\n\n{current_code}"
        reflect_history.append(types.Content(role="user", parts=[types.Part.from_text(text=critique_prompt)]))
        
        # Use our resilient call wrapper here too
        reflect_response = safe_generate_content(
            model='gemini-2.5-pro',
            contents=reflect_history,
            config=reflect_config
        )
        feedback = reflect_response.text
        reflect_history.append(types.Content(role="model", parts=[types.Part.from_text(text=feedback)]))
        
        gen_history.append(types.Content(role="user", parts=[types.Part.from_text(text=f"Integrate this feedback to improve the code:\n\n{feedback}")]))
        
        # Good practice: add a tiny sleep between loop iterations to stay under rate-limits
        time.sleep(1)

    return current_code
```

```python
generation_system_prompt = "You are a Python programmer tasked with generating high quality Python code"
reflection_system_prompt = "You are Andrej Karpathy, an experienced computer scientist"
user_msg = "Generate a Python implementation of the Merge Sort algorithm"

# --- Run Agent & Print Final Output ---
final_response = run_gemini_reflection_agent(
    user_msg=user_msg,
    generation_system_prompt=generation_system_prompt,
    reflection_system_prompt=reflection_system_prompt,
    n_steps=2 
)

print("\n==============================")
print("==== FINAL OPTIMIZED CODE ====")
print("==============================\n")
print(final_response)
```
