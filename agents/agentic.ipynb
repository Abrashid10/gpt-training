```python
import json
def get_current_weather(location: str, unit: str):
  if location == 'Madrid':
    return json.dumps({'location': location, "unit": f' 25 {unit}'})
  else:
    return json.dumps({'location': 'location not found', 'unit': unit})
  
get_current_weather(location='Madrid', unit='celsius')
```

```python
TOOL_SYSTEM_PROMPT = """
You are a function calling AI model. You are provided with function signatures within <tools></tools> XML tags. 
You may call one or more functions to assist with the user query. Don't make assumptions about what values to plug 
into functions. Pay special attention to the properties 'types'. You should use those types as in a Python dict.
For each function call return a json object with function name and arguments within <tool_call></tool_call> XML tags as follows:

<tool_call>
{"name": "get_current_weather", "arguments": {"location": "Madrid", "unit": "celsius"}}
</tool_call>

Here are the available tools:

{
    "name": "get_current_weather",
    "description": "Get the current weather in a given location",
    "parameters": {
        "type": "object",
        "properties": {
            "location": {
                "type": "string",
                "description": "The city and state, e.g. Madrid, Barcelona"
            },
            "unit": {
                "type": "string",
                "enum": ["celsius", "fahrenheit"],
                "description": "The temperature unit"
            }
        },
        "required": ["location"]
    }
}
"""
tool_chat_history = [
    {
        "role": "system",
        "content": TOOL_SYSTEM_PROMPT
    }
]
agent_chat_history = []

user_msg = {
    "role": "user",
    "content": "What's the current temperature in Madrid, in Celsius?"
}

tool_chat_history.append(user_msg)
agent_chat_history.append(user_msg)
```

```python
import re


def parse_tool_call(llm_response):
    # Extract the JSON from <tool_call> tags
    match = re.search(r'<tool_call>(.*?)</tool_call>', llm_response, re.DOTALL)
    if match:
        tool_call = json.loads(match.group(1))
        return tool_call["name"], tool_call["arguments"]
    return None, None

def execute_function(function_name, arguments):
    if function_name == "get_current_weather":
        return get_current_weather(**arguments)
```

```python
parsed_output_name, parsed_output_arg = parse_tool_call(output)
parsed_output_name, parsed_output_arg
```

```python
print(f"Name: {parsed_output_name}")
print(f"Args: {parsed_output_arg}")
```

```python
result = execute_function(parsed_output_name, parsed_output_arg)
result
```

----> ReAct Agent overall summary
-Thought: The model analyzes the current situation and decides what it needs to know or do next.Action: -The model picks a specific tool and runs it (for example, searching the web or executing code).-Observation: The model looks at the raw result returned by that tool.

```python
import math

@tool
def sum_two_elements(a: int, b: int) -> int:
    """
    Computes the sum of two integers.

    Args:
        a (int): The first integer to be summed.
        b (int): The second integer to be summed.

    Returns:
        int: The sum of `a` and `b`.
    """
    return a + b


@tool
def multiply_two_elements(a: int, b: int) -> int:
    """
    Multiplies two integers.

    Args:
        a (int): The first integer to multiply.
        b (int): The second integer to multiply.

    Returns:
        int: The product of `a` and `b`.
    """
    return a * b

@tool
def compute_log(x: int) -> float | str:
    """
    Computes the logarithm of an integer `x` with an optional base.

    Args:
        x (int): The integer value for which the logarithm is computed. Must be greater than 0.

    Returns:
        float: The logarithm of `x` to the specified `base`.
    """
    if x <= 0:
        return "Logarithm is undefined for values less than or equal to 0."
    
    return math.log(x)


available_tools = {
    "sum_two_elements": sum_two_elements,
    "multiply_two_elements": multiply_two_elements,
    "compute_log": compute_log
}
```

```python
print("Tool name: ", sum_two_elements.name)
display_markdown(sum_two_elements.fn_signature, raw = True)
```

```python
tools_signature = sum_two_elements.fn_signature + ",\n" + multiply_two_elements.fn_signature + ",\n" + compute_log.fn_signature
```

```python
REACT_SYSTEM_PROMPT = """
You are a function calling AI model. You operate by running a loop with the following steps: Thought, Action, Observation.
You are provided with function signatures within <tools></tools> XML tags.
You may call one or more functions to assist with the user query. Don' make assumptions about what values to plug
into functions. Pay special attention to the properties 'types'. You should use those types as in a Python dict.

For each function call return a json object with function name and arguments within <tool_call></tool_call> XML tags as follows:

<tool_call>
{"name": <function-name>,"arguments": <args-dict>, "id": <monotonically-increasing-id>}
</tool_call>

Here are the available tools / actions:

<tools> 
%s
</tools>

Example session:

<question>What's the current temperature in Madrid?</question>
<thought>I need to get the current weather in Madrid</thought>
<tool_call>{"name": "get_current_weather","arguments": {"location": "Madrid", "unit": "celsius"}, "id": 0}</tool_call>

You will be called again with this:

<observation>{0: {"temperature": 25, "unit": "celsius"}}</observation>

You then output:

<response>The current temperature in Madrid is 25 degrees Celsius</response>

Additional constraints:

- If the user asks you something unrelated to any of the tools above, answer freely enclosing your answer with <response></response> tags.
"""
```

```python
@tool
def write_str_to_txt(string_data: str, txt_filename: str):
    """
    Writes a string to a txt file.

    This function takes a string and writes it to a text file. If the file already exists, 
    it will be overwritten with the new data.

    Args:
        string_data (str): The string containing the data to be written to the file.
        txt_filename (str): The name of the text file to which the data should be written.
    """
    # Write the string data to the text file
    with open(txt_filename, mode='w', encoding='utf-8') as file:
        file.write(string_data)

    print(f"Data successfully written to {txt_filename}")
```

```python
from crewai import Agent, Task, Crew
from crewai import LLM
import asyncio
import nest_asyncio
import os
from crewai.tools import tool
from crewai.tools import tool

import crewai.llms.cache as _crewai_cache
_crewai_cache.mark_cache_breakpoint = lambda msg: msg


@tool("Write String to Text File")
def write_str_to_txt(content: str, filename: str = "./poem.txt") -> str:
    """
    Writes a string content to a text file.
    Args:
        content (str): The text content to write.
        filename (str): The path to the file. Defaults to './poem.txt'.
    Returns:
        str: A confirmation message.
    """
    try:
        with open(filename, "w", encoding="utf-8") as f:
            f.write(content)
        return f"Successfully saved to file: '{filename}'"
    except Exception as e:
        return f"Error writing to file: {str(e)}"

# Configure the LLM
llm = LLM(
    model="groq/llama-3.1-8b-instant",
    api_key=os.getenv("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1",
    # cache=False,
    # no_cache=True,
    custom_llm_provider="groq",
    # # Additional parameters to suppress cache_breakpoint
    # metadata={"cache": False}
    extra_headers={"no-cache": "true"}
)

# Create agents with CrewAI syntax
poet_agent = Agent(
    role="Poet",
    goal="Create beautiful poetry about the meaning of life",
    backstory="You are a well-known poet, who enjoys creating high quality poetry.",
    llm=llm,
    allow_delegation=False
)

translator_agent = Agent(
    role="Translator",
    goal="Translate poetry into Spanish",
    backstory="You are an expert translator especially skilled in Spanish",
    llm=llm,
    allow_delegation=False
)

writer_agent = Agent(
    role="Writer",
    goal="Write poems into txt files",
    backstory="You are an expert transcriber, that loves writing poems into txt files",
    llm=llm,
    allow_delegation=False,
    tools=[write_str_to_txt]  # Tools in CrewAI
)

# Define tasks
poem_task = Task(
    description="Write a poem about the meaning of life",
    expected_output="Just output the poem, without any title or introductory sentences",
    agent=poet_agent
)

translation_task = Task(
    description="Translate the poem into Spanish",
    expected_output="Just output the translated poem and nothing else",
    agent=translator_agent,
    context=[poem_task]  # Uses the output from poem_task
)

write_task = Task(
    description="Write the Spanish poem into './poem.txt' file",
    expected_output="A txt file containing the Spanish poem",
    agent=writer_agent,
    context=[translation_task]  # Uses the output from translation_task
)

# Create and run the crew
my_crew = Crew(
    agents=[poet_agent, translator_agent, writer_agent],
    tasks=[poem_task, translation_task, write_task],
    verbose=True,
    max_rpm=1
)

result = await my_crew.kickoff_async()
print(result)
```
